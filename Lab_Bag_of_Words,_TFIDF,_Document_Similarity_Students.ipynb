TFIDF consists of 2 steps: finding the TF and the IDF. The final result is just the product of TF and IDF.

```python
#this is a function for computing term frequency
def computeTF(wordDict, bagOfWords):
    tfDict = {}
    bagOfWordsCount = len(bagOfWords)#finds the length of list bagOfWords
    for word, count in wordDict.items():
        tfDict[word] = count / float(bagOfWordsCount) #finding term frequency
    return tfDict
```

```python
#term frquency for doc1
tfA = computeTF(vectorA, bowDOC1)
#find the term frequencies for doc2 and doc3, store them in variables tfB and tfC
tfB = computeTF(vectorB, bowDOC2)
tfC = computeTF(vectorC, bowDOC3)
```

```python
tfB
```

```python
#function for computing inverse document frequency
def computeIDF(documents):
    import math
    N = len(documents)

    idfDict = dict.fromkeys(documents[0].keys(), 0)
    for document in documents:
        for word, val in document.items():
            if val > 0:
                idfDict[word] += 1

    for word, val in idfDict.items():
        idfDict[word] = math.log(N / float(val))
    return idfDict
```

```python
idfs = computeIDF([vectorA, vectorB, vectorC])
idfs
```

```python
#tfidf is just tf*idf
def computeTFIDF(tfBagOfWords, idfs):
    tfidf = {}
    for word, val in tfBagOfWords.items():
        tfidf[word] = val * idfs[word]
    return tfidf
```

```python
#find the tfidf for 3 documents and represent them in a data frame
#your code here
tfidfA = computeTFIDF(tfA, idfs)
tfidfB = computeTFIDF(tfB, idfs)
tfidfC = computeTFIDF(tfC, idfs)
df = pd.DataFrame([tfidfA, tfidfB, tfidfC])
print(df)
```

- Cosine similarity is a metric used to measure how similar 2 documents are
- Mathematically, it measures the cosine of the angle between two vectors projected in a multi-dimensional space.

#Jaccard Similarity

- Jaccard Similarity is also known as the Jaccard index and Intersection over Union
- used to determine the similarity between two text document in terms of their context
- similarity is in terms of how many common words are exist over total words

![Image](https://miro.medium.com/max/744/1*XiLRKr_Bo-VdgqVI-SvSQg.png)

More here: https://en.wikipedia.org/wiki/Jaccard_index

```python
# Let's have 2 documents

doc1 = 'A is the brother of B'
doc2 = 'B is the friend of C who is not a brother of A'
```

```python
#convert documents to lower case - basic preprocessing!
doc1 = doc1.lower()
doc2 = doc2.lower()
```

```python
#split both the documents into tokens. Make sure that you have no duplicates. You might want to use sets for this purpose!?
#your code here
tokens1=set(doc1.split())
tokens2=set(doc2.split())
print(tokens1)
print(tokens2)
```

```python
#find common words from the 2 documents
#your code here
common=tokens1.intersection(tokens2)
common
```

```python
#find the vocabulary, total unique words in both documents
#your code here
unique_words=tokens1.union(tokens2)
unique_words
```

```python
#find jaccard similarity
#your code here
jaccard_similarity=len(common)/len(unique_words)
jaccard_similarity
```

#Task

Now you are given 4 documents.

- Preprocess them(lowercase, lemmatisation, stop word removal)

- Apply tfidf to convert the sentences into vectors

- Find which of the 4 documents are the most similar to each other


Your 4 sentences are:

1)"Hindi cinema, often known as Bollywood and formerly as Bombay cinema, is the Indian Hindi-language film industry based in Mumbai."

2)"films adapted from comic books have had plenty of success , whether they're about superheroes ( batman , superman , spawn ) , or geared toward kids ( casper ) or the arthouse crowd ( ghost world ) , but there's never really been a comic book like from hell before . "

3)"every now and then a movie comes along from a suspect studio , with every indication that it will be a stinker , and to everybody's surprise ( perhaps even the studio ) the film becomes a critical darling . "

4)"The Year 2000 problem, also known as the Y2K problem, the Millennium bug, Y2K bug, the Y2K glitch, or Y2K, refers to events related to the formatting and storage of calendar data for dates beginning in the year 2000"


```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import string

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

# Define the four documents
doc1 = "Hindi cinema, often known as Bollywood and formerly as Bombay cinema, is the Indian Hindi-language film industry based in Mumbai."
doc2 = "films adapted from comic books have had plenty of success , whether they're about superheroes ( batman , superman , spawn ) , or geared toward kids ( casper ) or the arthouse crowd ( ghost world ) , but there's never really been a comic book like from hell before . "
doc3 = "every now and then a movie comes along from a suspect studio , with every indication that it will be a stinker , and to everybody's surprise ( perhaps even the studio ) the film becomes a critical darling . "
doc4 = "The Year 2000 problem, also known as the Y2K problem, the Millennium bug, Y2K bug, the Y2K glitch, or Y2K, refers to events related to the formatting and storage of calendar data for dates beginning in the year 2000"

docs = [doc1, doc2, doc3, doc4]

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
punctuations = set(string.punctuation)

def preprocess(text):
    # Lowercase
    text = text.lower()
    # Tokenize
    tokens = word_tokenize(text)
    # Filter punctuation, stop words, and lemmatize
    cleaned_tokens = [
        lemmatizer.lemmatize(word) for word in tokens
        if word not in stop_words and word not in punctuations and word.isalnum()
    ]
    return " ".join(cleaned_tokens)

# Preprocess all documents
preprocessed_docs = [preprocess(doc) for doc in docs]

# Apply TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(preprocessed_docs)

# Calculate cosine similarity
similarity_matrix = cosine_similarity(tfidf_matrix)

# Create a dataframe for visual comparison
df_similarity = pd.DataFrame(
    similarity_matrix,
    columns=['Doc 1', 'Doc 2', 'Doc 3', 'Doc 4'],
    index=['Doc 1', 'Doc 2', 'Doc 3', 'Doc 4']
)

display(df_similarity)

# Find the most similar pair (excluding self-similarity of 1.0)
max_sim = -1
most_similar_pair = (None, None)

for i in range(len(docs)):
    for j in range(i + 1, len(docs)):
        sim = similarity_matrix[i][j]
        if sim > max_sim:
            max_sim = sim
            most_similar_pair = (i + 1, j + 1)

print(f"\nThe most similar documents are Document {most_similar_pair[0]} and Document {most_similar_pair[1]} with a Cosine Similarity score of {max_sim:.4f}.")
```
