#Neural Networks

Neural networks are multi-layer networks of neuronsthat we use to classify things, make predictions, etc. Below is the diagram of a simple neural network with one input layer, one output layer and one hidden layers of neuron.

![Image](https://ars.els-cdn.com/content/image/3-s2.0-B9780128115299000017-f01-07-9780128115299.jpg)

Now, we will design and view a simple neural network using Keras.

Keras is an open-source software library that provides a Python interface for artificial neural networks.

Now, the first thing is the dataset. We would use the scikit learn library to create a dataset for ourselves. The dataset would have 10 columns and 2 target classes. So, we would be training our classifier to learn to distinguish between 2 classes.

```python
from sklearn.datasets import make_classification
import pandas as pd
```

```python
#Creating the dataset
X, y = make_classification(n_samples = 500,
                                       n_features = 10,
                                       n_informative = 5,
                                       n_redundant = 0,
                                       n_classes = 2,
                                       weights = [.2, .3])
```

```python
#Viewing the dataset
print("Dataset: ");
print(pd.DataFrame(X, columns=["Feature 1", "Feature 2", "Feature 3", "Feature 4", "Feature 5","Feature 6", "Feature 7", "Feature 8", "Feature 9", "Feature 10"]).head())
```

#Splitting into train and test set



```python
#your task is to split the dataset such that there is 80% training data and 20% test data
from sklearn.model_selection import train_test_split
#your code here
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20)
```

#Designing the Neural Network

```python

# Importing the Keras libraries and packages
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.utils import plot_model
```

```python
# Initialising the Neural Network
model = Sequential()
```

```python
# Adding the input layer and the first hidden layer
model.add(Dense(units = 10, kernel_initializer = 'uniform', activation = 'relu', input_dim = 10))

# Adding the second hidden layer
model.add(Dense(units = 16, kernel_initializer = 'uniform', activation = 'relu'))

# Adding the output layer
model.add(Dense(units = 1, kernel_initializer = 'uniform', activation = 'sigmoid'))

# Compiling the ANN
model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])

plot_model(model, to_file='model_plot.png', show_shapes=True, show_layer_names=True)
```

```python
# Fitting the Neural Network to the Training set
model.fit(X_train, y_train, batch_size = 10, epochs = 100)
```

- How do you think activation functions affect Neural Networks?

Thinking of a neural network as a massive team of people passing messages down a line. Without activation functions, every person in that line would just multiply the incoming message by a simple number and pass it along. No matter how many people are in line, it would just boil down to one basic math equation (a straight line).

Activation functions are the decision-makers. They look at the incoming signal (

- How do you think tweaking learning rates affect Neural Networks? Would you choose a very high learning rate? If yes, why? Also, if no, why?

Imagine you are trying to find the lowest point of a dark, foggy valley (the lowest loss/error) while wearing blindfolds. The learning rate is the size of the steps you take with each movement.

If your step size is too small: You move like a snail. It takes forever to get anywhere, and you might get stuck in a tiny dip before reaching the bottom.

If your step size is huge: You take giant, wild leaps across the valley. One second you're on the left ridge, the next you vault clear over the bottom and end up high up on the right ridge, bouncing back and forth infinitely without ever settling down.

Would you choose a very high learning rate?
No.

Why not? Because a very high learning rate makes your training steps way too big. Instead of smoothly walking down into the lowest error rate, your model's weights will bounce around chaotically, overshoot the optimal solution, or diverge completely, meaning it will never actually learn anything useful.
